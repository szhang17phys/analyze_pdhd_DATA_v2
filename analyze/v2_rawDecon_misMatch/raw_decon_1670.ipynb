#### 1670 files (after TPC cut); For Raw and Decon:

```python
import matplotlib.pyplot as plt
import numpy as np

# Function to read values from a text file into a list
def read_values_from_file(file_name):
    with open(file_name, "r") as file:
        values = [int(line.strip()) for line in file]
    return values

# Read values from both text files
raw_values = read_values_from_file("./1670raw_extract.txt")
decon_values = read_values_from_file("./1670decon_extract.txt")

print("Entries of Raw  : ", len(raw_values))
print("Entries of Decon: ", len(decon_values))


# Plotting the histogram (X distribution)---------------
plt.figure(figsize=(10,3), dpi=400)

# Define the range and bin width
bin_width = 1
range_min = 0
range_max = 125
bins = np.arange(range_min, range_max + bin_width, bin_width)

# Create the histograms
raw_hist, _ = np.histogram(raw_values, bins=bins, range=(range_min, range_max))
decon_hist, _ = np.histogram(decon_values, bins=bins, range=(range_min, range_max))

# Plot the histograms
plt.hist(raw_values, bins=bins, range=(range_min, range_max), edgecolor='red', 
         histtype='step', linewidth=0.7, label="Raw")
plt.hist(decon_values, bins=bins, range=(range_min, range_max), edgecolor='blue', 
         histtype='stepfilled', alpha=0.6, color='blue', linewidth=0.7, label="Decon")

# Customize the plot
plt.xlim(0, 125)
plt.xlabel('opch', labelpad=2, fontsize=8)
plt.ylabel('#Counts', labelpad=1, fontsize=8)
plt.tick_params(labelsize=7)  # Set the size of ticks
plt.grid(color='gray', linestyle='--', linewidth=0.2, zorder=0)
plt.legend(loc='upper center', bbox_to_anchor=(0.2, 0.89), fontsize=11, markerscale=6, ncol=1, prop={'size': 7})

# Print Raw and Decon values separately and the difference
print("\nRaw bin contents:")
for i in range(len(raw_hist)):
    print(f"Bin {i + 1}: Raw = {raw_hist[i]}")

print("\nDecon bin contents:")
for i in range(len(decon_hist)):
    print(f"Bin {i + 1}: Decon = {decon_hist[i]}")

print("\nBin differences (Raw - Decon):")
for i in range(len(raw_hist)):
    difference = raw_hist[i] - decon_hist[i]
    print(f"Bin {i + 1}: Difference = {difference}")

# Show the plot
plt.show()
```
