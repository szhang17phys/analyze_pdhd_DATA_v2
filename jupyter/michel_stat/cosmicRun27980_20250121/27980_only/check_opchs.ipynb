### EXPECTED OpChs for Events before / after TPC cuts:

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

import os


# Path to the output file produced earlier
initial_opch_file = "./opch_explore/opchs_initial.txt"
filtered_opch_file = './opch_explore/opchs_filtered.txt'

# Read all OpCh values into a 1D list
opch_initial = []
opch_filtered = []


with open(initial_opch_file, "r") as f:
    for line in f:
        line = line.strip()
        if line:
            opch_initial.append(int(line))

with open(filtered_opch_file, "r") as f:
    for line in f:
        line = line.strip()
        if line:
            opch_filtered.append(int(line))


# Inspect result
print("Size of initial opchs : ", len(opch_initial))
print("Size of filtered opchs: ", len(opch_filtered))



```

```python
# Plotting the histogram (X distribution)---------------
plt.figure(figsize=(4,3),dpi=400)

#Define the range and bin width
bin_width = 1
range_min = 0
range_max = 160
bins = np.arange(range_min, range_max + bin_width, bin_width)

#plt.plot([4, 4], [0, 15000], color="magenta", linewidth=1.5, linestyle='--')

plt.hist(opch_initial, bins=bins, range=(range_min, range_max), edgecolor='black', 
         histtype='step', linewidth=0.8, label='Before ANY Cut')
plt.hist(opch_filtered, bins=bins, range=(range_min, range_max), edgecolor='red', 
         histtype='step', linewidth=0.8, label='After TPC Cut')

plt.legend(loc='upper center', bbox_to_anchor=(0.75, 0.95), fontsize=10, markerscale=6, ncol=1, prop={'size': 7})

plt.xlim(0, 160)
plt.xlabel('Expected Optical Channel Label', labelpad=2, fontsize=8)
plt.ylabel('#Counts', labelpad=1, fontsize=8)
plt.ylim(0.1, 150000)
#plt.yscale('log')
plt.title("Not Very Helpful ...")
plt.tick_params(labelsize=7)#set the size of ticks---
plt.grid(color='gray', linestyle='--', linewidth=0.2, zorder=0)

plt.show() 
```
