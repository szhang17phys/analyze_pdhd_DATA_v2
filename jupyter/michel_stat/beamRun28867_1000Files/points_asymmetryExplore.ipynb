##  Decay point distribution drawing; for 1000 FILES; beam run 28867

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D



# Specify the file name
#read data---
decayX = []
decayY = []
decayZ = []
michelScore = []
michelHits = []
startX = []
startY = []
startZ = []


with open('./statResult/decayX.txt', 'r') as filehandle:
    for line in filehandle:
        currentPlace = line[:-1]
        decayX.append(float(currentPlace))
with open('./statResult/decayY.txt', 'r') as filehandle:
    for line in filehandle:
        currentPlace = line[:-1]
        decayY.append(float(currentPlace))
with open('./statResult/decayZ.txt', 'r') as filehandle:
    for line in filehandle:
        currentPlace = line[:-1]
        decayZ.append(float(currentPlace))
        
with open('./statResult/michelScore.txt', 'r') as filehandle:
    for line in filehandle:
        currentPlace = line[:-1]
        michelScore.append(float(currentPlace))
with open('./statResult/michelHits.txt', 'r') as filehandle:
    for line in filehandle:
        currentPlace = line[:-1]
        michelHits.append(float(currentPlace))        

with open('./statResult/startX.txt', 'r') as filehandle:
    for line in filehandle:
        currentPlace = line[:-1]
        startX.append(float(currentPlace))
with open('./statResult/startY.txt', 'r') as filehandle:
    for line in filehandle:
        currentPlace = line[:-1]
        startY.append(float(currentPlace))
with open('./statResult/startZ.txt', 'r') as filehandle:
    for line in filehandle:
        currentPlace = line[:-1]
        startZ.append(float(currentPlace))



print("Length of decayX: ", len(decayX))
print("Length of decayY: ", len(decayY))
print("Length of decayZ: ", len(decayZ))
print("Length of michelScore: ", len(michelScore))
print("Length of michelHits: ", len(michelHits))
print("Length of startX: ", len(startX))
print("Length of startY: ", len(startY))
print("Length of startZ: ", len(startZ))
```

## Apply Cuts Manually:

```python

#Space Cut---
ScutX = []#End point---
ScutY = []
ScutZ = []
ScutX_start = []#starting point---
ScutY_start = []
ScutZ_start = []
ScutMS = []
ScutMH = []

#Michel Score cut (after space cut)---
MScutX = []
MScutY = []
MScutZ = []
MScutX_start = []
MScutY_start = []
MScutZ_start = []
MScutMS = []
MScutMH = []

#Michel Hits cut (after michel score cut)---
MHcutX = []
MHcutY = []
MHcutZ = []
MHcutX_start = []
MHcutY_start = []
MHcutZ_start = []
MHcutMS = []
MHcutMH = []


#Selection------
for i in range(0, len(decayX)):
    X = decayX[i]
    Y = decayY[i]
    Z = decayZ[i]
    X_start = startX[i]
    Y_start = startY[i]
    Z_start = startZ[i]
    MS = michelScore[i]
    MH = michelHits[i]
  
    #Default region: 0<y<610, 0<z<465 (SPACE CUT)---
    if -356 < X < 356 and 30 < Y < 580 and 30 < Z < 435:
        ScutX.append(X)
        ScutY.append(Y)
        ScutZ.append(Z)
        ScutX_start.append(X_start)
        ScutY_start.append(Y_start)
        ScutZ_start.append(Z_start)
        ScutMS.append(MS)
        ScutMH.append(MH)      

        if 0.4 < MS: #(Michel score cut)---
            MScutX.append(X)
            MScutY.append(Y)
            MScutZ.append(Z)
            MScutX_start.append(X_start)
            MScutY_start.append(Y_start)
            MScutZ_start.append(Z_start)
            MScutMS.append(MS)
            MScutMH.append(MH)  
            
            if 5 < MH: #Michel hits cut---
                MHcutX.append(X)
                MHcutY.append(Y)
                MHcutZ.append(Z)
                MHcutX_start.append(X_start)
                MHcutY_start.append(Y_start)
                MHcutZ_start.append(Z_start)
                MHcutMS.append(MS)
                MHcutMH.append(MH)
            

#Initial size of dataset:-----
print("Size of X (initial) : ", len(decayX))
print("Size of Y (initial) : ", len(decayY))
print("Size of Z (initial) : ", len(decayZ))
print("Size of X_start (initial) : ", len(startX))
print("Size of Y_start (initial) : ", len(startY))
print("Size of Z_start (initial) : ", len(startZ))
print("Size of MS (initial): ", len(michelScore))
print("Size of MH (initial): ", len(michelHits))

print("\nSize of X (space cut) : ", len(ScutX), "({:.3f}%)".format(len(ScutX)/len(decayX)*100))
print("Size of Y (space cut) : ", len(ScutY))
print("Size of Z (space cut) : ", len(ScutZ))
print("Size of X_start (space cut) : ", len(ScutX_start))
print("Size of Y_start (space cut) : ", len(ScutY_start))
print("Size of Z_start (space cut) : ", len(ScutZ_start))
print("Size of MS (space cut): ", len(ScutMS))
print("Size of MH (space cut): ", len(ScutMH))

print("\nSize of X (space cut & MS cut) : ", len(MScutX), "({:.3f}%)".format(len(MScutX)/len(decayX)*100))
print("Size of Y (space cut & MS cut) : ", len(MScutY))
print("Size of Z (space cut & MS cut) : ", len(MScutZ))
print("Size of X_start (space cut & MS cut) : ", len(MScutX_start))
print("Size of Y_start (space cut & MS cut) : ", len(MScutY_start))
print("Size of Z_start (space cut & MS cut) : ", len(MScutZ_start))
print("Size of MS (space cut & MS cut): ", len(MScutMS))
print("Size of MH (space cut & MS cut): ", len(MScutMH))


print("\nSize of X (space cut & MS cut & MH cut) : ", len(MHcutX), "({:.3f}%)".format(len(MHcutX)/len(decayX)*100))
print("Size of Y (space cut & MS cut & MH cut) : ", len(MHcutY))
print("Size of Z (space cut & MS cut & MH cut) : ", len(MHcutZ))
print("Size of X_start (space cut & MS cut & MH cut) : ", len(MHcutX_start))
print("Size of Y_start (space cut & MS cut & MH cut) : ", len(MHcutY_start))
print("Size of Z_start (space cut & MS cut & MH cut) : ", len(MHcutZ_start))
print("Size of MS (space cut & MS cut & MH cut): ", len(MHcutMS))
print("Size of MH (space cut & MS cut & MH cut): ", len(MHcutMH))



```

###  Explore asymmetry of X didtribution:

```python
leftSX = [] #left start X---
leftSY = [] #left start Y---
leftSZ = [] #left start Z---
leftDX = [] #left decay X---
leftDY = []
leftDZ = []

rightSX = [] #left start X---
rightSY = [] #left start Y---
rightSZ = [] #left start Z---
rightDX = [] #left decay X---
rightDY = []
rightDZ = []




#Events located at APA2/4:
MHcutX_start_apa = []
MHcutX_apa = []
MHcutY_start_apa = []
MHcutY_apa = []
MHcutZ_start_apa = []
MHcutZ_apa = []

for i in range(0, len(MHcutX)):
    sX = MHcutX_start[i]
    eX = MHcutX[i]
    sY = MHcutY_start[i]
    eY = MHcutY[i]
    sZ = MHcutZ_start[i]
    eZ = MHcutZ[i]

    #Constrain ending point as well as starting point!!!
    if eX > 0 and 232.5 < eZ < 465 and sX < 0 and 232.5 < sZ < 465:
        MHcutX_start_apa.append(sX)
        MHcutX_apa.append(eX)
        MHcutY_start_apa.append(sY)
        MHcutY_apa.append(eY)
        MHcutZ_start_apa.append(sZ)
        MHcutZ_apa.append(eZ)





for i in range(0, len(MHcutX_apa)):
#for i in range(0, len(MHcutX)):    
    # sX = MHcutX_start[i]
    # eX = MHcutX[i]
    # sY = MHcutY_start[i]
    # eY = MHcutY[i]
    # sZ = MHcutZ_start[i]
    # eZ = MHcutZ[i]
    
    sX = MHcutX_start_apa[i]
    eX = MHcutX_apa[i]
    sY = MHcutY_start_apa[i]
    eY = MHcutY_apa[i]
    sZ = MHcutZ_start_apa[i]
    eZ = MHcutZ_apa[i]

    if sX <= eX:
        leftSX.append(sX)
        leftSY.append(sY)
        leftSZ.append(sZ)
        leftDX.append(eX)
        leftDY.append(eY)
        leftDZ.append(eZ)
    else:
        rightSX.append(sX)
        rightSY.append(sY)
        rightSZ.append(sZ)
        rightDX.append(eX)
        rightDY.append(eY)
        rightDZ.append(eZ)        

print("For single APA region!--------------------------")
print("# of left-emitted particles : ", len(leftSX))
print("# of right-emitted particles: ", len(rightSX))
print("Total particles (left + right): ", len(leftSX)+len(rightSX))
print("Initial particles: ", len(MHcutX))



#Z-X plane ploting (Top View)----------------------------------------
plt.figure(figsize=(4,3),dpi=400)

#auxiliary line, along y(geo) and x axis---
aux2 = [0, 465, 465, 0, 0] #y shoud be y---
aux1 = [356, 356, -356, -356, 356]
plt.plot(aux1, aux2, color="blue", linewidth=0.8, linestyle='-')


#Drawing of data points------
plt.scatter(MHcutX_apa, MHcutZ_apa, c='magenta', s=2.0, alpha=1.0, marker='.', label='Magenta Bias')


plt.axis('scaled')
plt.xlim(-450, 450)#z---
plt.ylim(-50, 515)
plt.xlabel('X axis [cm]', labelpad=1, fontsize=5)
plt.ylabel('Z axis [cm]', labelpad=1, fontsize=5)
plt.tick_params(labelsize=4)#set the size of ticks---
plt.grid(color='gray', linestyle='--', linewidth=0.3, zorder=0)
plt.show() 
```
