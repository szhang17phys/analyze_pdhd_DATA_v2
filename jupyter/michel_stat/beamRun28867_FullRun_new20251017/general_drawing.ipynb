```python
import numpy as np
import matplotlib.pyplot as plt

# Data
MS_cut = [0.4, 0.45, 0.5, 0.6, 0.7, 0.8]
tau_fit = [1883, 1860, 1831, 1823, 1895, 1803]
tau_unc = [75, 76, 78, 85, 99, 114]
purity = [87.2, 88.6, 90.4, 93.1, 95.0, 95.8]

# Figure setup
fig, ax1 = plt.subplots(figsize=(5, 4), dpi=400)

# --- Left y-axis: tau_fit with uncertainties ---
ax1.errorbar(
    MS_cut, tau_fit, yerr=tau_unc,
    fmt='o', color='blue', ecolor='blue',
    elinewidth=1, capsize=2, markersize=4, label=r'$\tau_{\rm fit}$'
)

ax1.axhline(
    y=1910, color='magenta', linestyle='--', linewidth=1,
    label=r'Theoretical Expected lifetime'
)


ax1.set_xlabel('Michel Score Cut')
ax1.set_ylabel(r'Fitted Lifetime $\tau_{\rm fit}$ (ns)', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')
ax1.set_xlim(0.3, 0.9)
ax1.set_ylim(1500, 2200)

plt.grid(color='gray', linestyle='--', linewidth=0.2, zorder=0)

# --- Right y-axis: purity ---
ax2 = ax1.twinx()
ax2.plot(
    MS_cut, purity, 's--', color='tab:red', markersize=4, label='Purity'
)
ax2.set_ylabel('Purity (%)', color='tab:red')
ax2.tick_params(axis='y', labelcolor='tab:red')
ax2.set_ylim(0, 120)

# --- Combined legend ---
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left', fontsize=7)



#plt.tight_layout()
plt.show()
```
