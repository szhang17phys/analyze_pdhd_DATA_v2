## TPC cuts:


### Apply cuts gradually:

```python
#Add cuts gradually:

trueE_initial = []
trueL_initial = []

#After all cuts: Space, MS, MH---
ScutX = []#After space cut---
ScutY = []
ScutZ = []
ScutMS = []
ScutMH = []
ScutE = []#energy---
ScutL = []#lifetime---

MScutX = []#After space & MS cuts---
MScutY = []
MScutZ = []
MScutMS = []
MScutMH = []
MScutE = []
MScutL = []

MHcutX = []#After space and MS, MH cuts---
MHcutY = []
MHcutZ = []
MHcutMS = []
MHcutMH = []
MHcutE = []
MHcutL = []







#Selection------
for i in range(0, len(decayX_initial)):
    X = decayX_initial[i]
    Y = decayY_initial[i]
    Z = decayZ_initial[i]
    MS = MS_initial[i]
    MH = MH_initial[i]
    E = energy_initial[i]
    L = lifetime_initial[i]
    if E > 0:
        trueE_initial.append(E)
    if L > 0:
        trueL_initial.append(L)
        
    #Default region: 0<y<610, 0<z<465 (SPACE CUT)---
    if -356 < X < 356 and 30 < Y < 580 and 30 < Z < 435:    
        ScutX.append(X)
        if E > 0:
            ScutE.append(E)
        if L > 0:
            ScutL.append(L)

        if 0.4 < MS: #(Michel score cut)---
            MScutX.append(X)
            if E > 0:
                MScutE.append(E)
            if L > 0:
                MScutL.append(L)
            
            if 5 <MH: #Michel hits cut---                
                MHcutX.append(X)
                MHcutY.append(Y)
                MHcutZ.append(Z)
                MHcutMS.append(MS)
                MHcutMH.append(MH)
                if E > 0:
                    MHcutE.append(E)
                if L > 0:
                    MHcutL.append(L)   
                    
                    
                    
                    
                    
                    
                    
 #Initial size of dataset:-----
print("Reco events (initial): ", len(decayX_initial))
print("True events (initial): ", len(trueE_initial), "({:.2f}%)".format(len(trueE_initial)/len(decayX_initial)*100))

print("\nReco events (Space cut): ", len(ScutX), "({:.2f}%)".format(len(ScutX)/len(decayX_initial)*100))
print("True events (Space cut): ", len(ScutE), "({:.2f}%)".format(len(ScutE)/len(ScutX)*100))

print("\nReco events (Space & MS cuts): ", len(MScutX), "({:.2f}%)".format(len(MScutX)/len(decayX_initial)*100))
print("True events (Space & MS cuts): ", len(MScutE), "({:.2f}%)".format(len(MScutE)/len(MScutX)*100))


print("\nReco events (Space & MS & MH cuts): ", len(MHcutX), "({:.2f}%)".format(len(MHcutX)/len(decayX_initial)*100))
print("True events (Space & MS & MH cuts): ", len(MHcutE), "({:.2f}%)".format(len(MHcutE)/len(MHcutX)*100))

print("\n\nSize of Y (space cut & MS cut & MH cut) : ", len(MHcutY))
print("Size of Z (space cut & MS cut & MH cut) : ", len(MHcutZ))
print("Size of MS (space cut & MS cut & MH cut): ", len(MHcutMS))
print("Size of MH (space cut & MS cut & MH cut): ", len(MHcutMH))




```
