```python
decayX_initial = []#Candidates after TPC cut---
decayY_initial = []
decayZ_initial = []

startX_initial = []#Candidates after TPC cut---
startY_initial = []
startZ_initial = []

MS_initial = []
MH_initial = []

energy_initial = []
lifetime_initial = []
pdg_initial = []


decayX_file_paths = [
    './statResult_20250828/decayX_New20250828_initial.txt'
]
for file_path in decayX_file_paths:
    with open(file_path, 'r') as filehandle:
        for line in filehandle:
            value = line.strip()
            if value:  # Skip empty lines safely
                decayX_initial.append(float(value))


decayY_file_paths = [
    './statResult_20250828/decayY_New20250828_initial.txt'
]
for file_path in decayY_file_paths:
    with open(file_path, 'r') as filehandle:
        for line in filehandle:
            value = line.strip()
            if value:  # Skip empty lines safely
                decayY_initial.append(float(value))
        

decayZ_file_paths = [
    './statResult_20250828/decayZ_New20250828_initial.txt'
]
for file_path in decayZ_file_paths:
    with open(file_path, 'r') as filehandle:
        for line in filehandle:
            value = line.strip()
            if value:  # Skip empty lines safely
                decayZ_initial.append(float(value))


startX_file_paths = [
    './statResult_20250828/startX_New20250828_initial.txt'
]
for file_path in startX_file_paths:
    with open(file_path, 'r') as filehandle:
        for line in filehandle:
            value = line.strip()
            if value:  # Skip empty lines safely
                startX_initial.append(float(value))


startY_file_paths = [
    './statResult_20250828/startY_New20250828_initial.txt'
]
for file_path in startY_file_paths:
    with open(file_path, 'r') as filehandle:
        for line in filehandle:
            value = line.strip()
            if value:  # Skip empty lines safely
                startY_initial.append(float(value))
        

startZ_file_paths = [
    './statResult_20250828/startZ_New20250828_initial.txt'
]
for file_path in startZ_file_paths:
    with open(file_path, 'r') as filehandle:
        for line in filehandle:
            value = line.strip()
            if value:  # Skip empty lines safely
                startZ_initial.append(float(value))


michelScore_file_paths = [
    './statResult_20250828/michelScore_New20250828_initial.txt'
]
for file_path in michelScore_file_paths:
    with open(file_path, 'r') as filehandle:
        for line in filehandle:
            value = line.strip()
            if value:  # Skip empty lines safely
                MS_initial.append(float(value))


michelHits_file_paths = [
    './statResult_20250828/michelHits_New20250828_initial.txt'
]
for file_path in michelHits_file_paths:
    with open(file_path, 'r') as filehandle:
        for line in filehandle:
            value = line.strip()
            if value:  # Skip empty lines safely
                MH_initial.append(float(value))


energy_file_paths = [
    './statResult_20250828/energy_New20250828_initial.txt'
]
for file_path in energy_file_paths:
    with open(file_path, 'r') as filehandle:
        for line in filehandle:
            value = line.strip()
            if value:  # Skip empty lines safely
                energy_initial.append(float(value))


lifetime_file_paths = [
    './statResult_20250828/lifetime_New20250828_initial.txt'
]
for file_path in lifetime_file_paths:
    with open(file_path, 'r') as filehandle:
        for line in filehandle:
            value = line.strip()
            if value:  # Skip empty lines safely
                lifetime_initial.append(float(value))

pdg_file_paths = [
    './statResult_20250828/pdg_New20250828_initial.txt'
]
for file_path in pdg_file_paths:
    with open(file_path, 'r') as filehandle:
        for line in filehandle:
            value = line.strip()
            if value:  # Skip empty lines safely
                pdg_initial.append(int(value))



print("Candidates ( MH > 1e-9): decayX: ", len(decayX_initial))
print("Candidates ( MH > 1e-9): decayY: ", len(decayY_initial))
print("Candidates ( MH > 1e-9): decayZ: ", len(decayZ_initial))
print("Candidates ( MH > 1e-9): startX: ", len(startX_initial))
print("Candidates ( MH > 1e-9): startY: ", len(startY_initial))
print("Candidates ( MH > 1e-9): startZ: ", len(startZ_initial))

print("Candidates ( MH > 1e-9): MS: ", len(MS_initial))
print("Candidates ( MH > 1e-9): MH: ", len(MH_initial))

print("Candidates ( MH > 1e-9): energy  : ", len(energy_initial))
print("Candidates ( MH > 1e-9): lifetime: ", len(lifetime_initial))

print("Candidates ( MH > 1e-9): pdg: ", len(pdg_initial))


posEnergy = sum(1 for x in energy_initial if x > 0)
print("Number of positive posEnergy  : ", posEnergy, "({:.3f}%)".format(posEnergy/len(energy_initial)*100))

posLifetime = sum(1 for x in lifetime_initial if x > 0)
print("Number of positive posLifetime: ", posLifetime, "({:.3f}%)".format(posLifetime/len(lifetime_initial)*100))

```

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit




# Seperate Michel e^- and e^+------------------------------
life_pM = []
life_nM = []

for i in range(0, len(pdg_initial)):
    pdg = pdg_initial[i]
    life = lifetime_initial[i]

    if pdg < 0:
        life_pM.append(life)
    if pdg > 0:
        life_nM.append(life)
    
count_0    = pdg_initial.count(0)

print("\nMichel e^+: ", len(life_pM), "/", len(pdg_initial), "({:.2f}%)".format(len(life_pM)/len(pdg_initial)*100))
print("Michel e^-: ", len(life_nM), "/", len(pdg_initial), "({:.2f}%)".format(len(life_nM)/len(pdg_initial)*100))
print("False     : ", count_0, "/", len(pdg_initial), "({:.2f}%)".format(count_0/len(pdg_initial)*100))








# Define the new exponential decay function=======================================================
# -------------------------
# Model: N1*exp(-t/tau1) + N2*exp(-t/2.2)  (tau2 fixed)
# -------------------------
def exp_decay(t, N1, tau1, N2):
    return N1 * np.exp(-t / tau1) + N2 * np.exp(-t / 2.1)

# Settings for the histogram
plt.figure(figsize=(4, 3), dpi=400)
bin_width = 0.05
range_min = -20
range_max = 20
bins = np.arange(range_min, range_max + bin_width, bin_width)

# Histogram the data
counts, bin_edges = np.histogram(lifetime_initial, bins=bins, range=(range_min, range_max))
bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

# Select data within the fitting range
fit_mask = (bin_centers >= 0.05) & (bin_centers <= 10)
x_fit_all = bin_centers[fit_mask]
y_fit_all = counts[fit_mask]

# --- Skip zero bins (avoid infinite weight from 1/sigma^2 with sigma=sqrt(y)) ---
nz = y_fit_all > 0
x_fit = x_fit_all[nz]
y_fit = y_fit_all[nz]
sigma = np.sqrt(y_fit)  # Poisson uncertainty per bin

# Initial guesses (safe)
N1_0 = float(np.max(y_fit)) if y_fit.size else 1.0
tau1_0 = 2.0
N2_0 = max(N1_0 / 10.0, 1.0)

# Weighted least squares fit (absolute_sigma=True → covariance reflects our sigmas)
popt, pcov = curve_fit(
    exp_decay, x_fit, y_fit,
    p0=[N1_0, tau1_0, N2_0],
    sigma=sigma,
    absolute_sigma=True,
    bounds=([0.0, 0.0, 0.0], [np.inf, np.inf, np.inf]),
    maxfev=20000,
)
N1_fit, tau1_fit, N2_fit = popt
perr = np.sqrt(np.diag(pcov))  # 1σ uncertainties
N1_err, tau1_err, N2_err = perr

# Compute chi-squared with provided sigmas
y_exp = exp_decay(x_fit, *popt)
residual = (y_fit - y_exp) / sigma
chi2 = np.sum(residual**2)
ndf = max(len(x_fit) - len(popt), 1)
chi2_ndf = chi2 / ndf

# Smooth curve for plotting
x_smooth = np.linspace(0.05, 10, 500)
y_smooth = exp_decay(x_smooth, *popt)

# Plot histogram (data)
plt.hist(lifetime_initial, bins=bins, range=(range_min, range_max),
         edgecolor='magenta', histtype='step', linewidth=0.8, label="Data")

# Plot the weighted-χ² fit
plt.plot(
    x_smooth, y_smooth, 'b-',
    label=(
        r'Fit: $N_1 e^{-t/\tau_1} + N_2 e^{-t/2.1}$' + '\n'
        + rf'$N_1={N1_fit:.1f}\pm{N1_err:.1f}$,  $\tau_1={tau1_fit:.3f}\pm{tau1_err:.3f}\ \mu s$' + '\n'
        + rf'$N_2={N2_fit:.1f}\pm{N2_err:.1f}$,  $\tau_2=2.1\ \mu s$ (fixed)' + '\n'
        + rf'$\chi^2/\mathrm{{ndf}} = {chi2:.1f}/{ndf}$'
    )
)

# Final plot settings (your style)
plt.xlim(-1, 18)
plt.ylim(0.1, 200000)
plt.yscale('log')
plt.xlabel('Lifetime [$\mu$s]', labelpad=2, fontsize=8)
plt.ylabel('#Counts', labelpad=1, fontsize=8)
plt.tick_params(labelsize=7)
plt.grid(color='gray', linestyle='--', linewidth=0.2, zorder=0)
plt.title('Lifetime Distribution of Michel True Events', fontsize=9)
plt.legend(fontsize=6)
plt.tight_layout()
plt.show()

# Optional: also print to console
print(
    f"N1 = {N1_fit:.3f} ± {N1_err:.3f}\n"
    f"tau1 = {tau1_fit:.5f} ± {tau1_err:.5f} us\n"
    f"N2 = {N2_fit:.3f} ± {N2_err:.3f}\n"
    f"chi2/ndf = {chi2:.2f}/{ndf} = {chi2_ndf:.3f}"
)











#Fitting seperating for Michel e^+ and e^-=============================================================

# --- single exponential ---
def exp1(t, N, tau):
    return N * np.exp(-t / tau)

# --- plotting / binning setup ---
plt.figure(figsize=(4, 3), dpi=400)
bin_width = 0.05
range_min, range_max = -20, 20
bins = np.arange(range_min, range_max + bin_width, bin_width)

# --- helper: weighted fit (σ = sqrt(y)), skip zeros, draw ---
def fit_and_draw(sample, label_prefix, fit_range=(0.05, 10)):
    # histogram
    counts, bin_edges = np.histogram(sample, bins=bins, range=(range_min, range_max))
    bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

    # choose fit window & drop zero bins
    m = (bin_centers >= fit_range[0]) & (bin_centers <= fit_range[1])
    x_all = bin_centers[m]
    y_all = counts[m]
    nz = y_all > 0
    x_fit = x_all[nz]
    y_fit = y_all[nz]

    # abort gracefully if too few points
    if len(y_fit) < 3:
        plt.hist(sample, bins=bins, range=(range_min, range_max),
                 histtype='step', linewidth=0.8, label=f"{label_prefix} MC data")
        return (np.nan, np.nan, np.nan, np.nan, np.nan)

    sigma = np.sqrt(y_fit)

    # safe initial guesses & bounds
    N0, tau0 = float(np.max(y_fit)), 2.0
    bounds = ([0.0, 1e-6], [np.inf, np.inf])

    # weighted least squares
    popt, pcov = curve_fit(
        exp1, x_fit, y_fit, p0=[N0, tau0],
        sigma=sigma, absolute_sigma=True,
        bounds=bounds, maxfev=20000
    )
    N_fit, tau_fit = popt
    N_err, tau_err = np.sqrt(np.diag(pcov))

    # chi2/ndf
    y_mod = exp1(x_fit, *popt)
    r = (y_fit - y_mod) / sigma
    chi2 = float(np.sum(r**2))
    ndf = max(len(x_fit) - len(popt), 1)
    chi2_ndf = chi2 / ndf

    # smooth curve
    x_smooth = np.linspace(fit_range[0], fit_range[1], 500)
    y_smooth = exp1(x_smooth, *popt)

    # draw histogram (step)
    plt.hist(sample, bins=bins, range=(range_min, range_max),
             histtype='step', linewidth=0.8, label=f"{label_prefix} data")

    # draw fit
    plt.plot(
        x_smooth, y_smooth,
        label=(fr"{label_prefix} fit: $Ne^{{-t/\tau}}$" + "\n"
               + rf"$N={N_fit:.1f}\pm{N_err:.1f}$,  $\tau={tau_fit:.3f}\pm{tau_err:.3f}\ \mu s$" + "\n"
               + rf"$\chi^2/\mathrm{{ndf}} = {chi2:.1f}/{ndf}$")
    )

    return (N_fit, tau_fit, N_err, tau_err, chi2_ndf)

# ---- run for both lists ----
N_p, tau_p, dN_p, dtau_p, chi2ndf_p = fit_and_draw(np.asarray(life_pM), "e⁺")
N_n, tau_n, dN_n, dtau_n, chi2ndf_n = fit_and_draw(np.asarray(life_nM), "e⁻")

# --- final plot cosmetics (kept similar to your style) ---
plt.xlim(-2, 18)
plt.ylim(0.1, 1000)
plt.yscale('log')
plt.xlabel('Lifetime [$\mu$s]', labelpad=2, fontsize=8)
plt.ylabel('#Counts', labelpad=1, fontsize=8)
plt.tick_params(labelsize=7)
plt.grid(color='gray', linestyle='--', linewidth=0.2, zorder=0)
plt.title('Lifetime Distributions: Michel $e^+$ vs $e^-$', fontsize=9)
plt.legend(fontsize=6)
plt.tight_layout()
plt.show()


```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit




# Seperate Michel e^- and e^+------------------------------
TPC_life_pM = []
TPC_life_nM = []

for i in range(0, len(TPC_pdg)):
    pdg = TPC_pdg[i]
    life = TPC_lifetime[i]

    if pdg < 0:
        TPC_life_pM.append(life)
    if pdg > 0:
        TPC_life_nM.append(life)
    
TPC_count_0    = TPC_pdg.count(0)

print("\nMichel e^+: ", len(TPC_life_pM), "/", len(TPC_pdg), "({:.2f}%)".format(len(TPC_life_pM)/len(TPC_pdg)*100))
print("Michel e^-: ", len(TPC_life_nM), "/", len(TPC_pdg), "({:.2f}%)".format(len(TPC_life_nM)/len(TPC_pdg)*100))
print("False     : ", TPC_count_0, "/", len(TPC_pdg), "({:.2f}%)".format(TPC_count_0/len(TPC_pdg)*100))








# Define the new exponential decay function=======================================================
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# ============================================================
# Exponential decay model (bin-by-bin, tau2 fixed)
# ============================================================
def exp_decay(t, N1, tau1, N2):
    return N1 * np.exp(-t / tau1) + N2 * np.exp(-t / 2.1)

# ============================================================
# Histogram setup
# ============================================================
plt.figure(figsize=(4, 3), dpi=400)
bin_width = 0.05
range_min = -20
range_max = 20
bins = np.arange(range_min, range_max + bin_width, bin_width)

# Histogram the data
counts, bin_edges = np.histogram(TPC_lifetime, bins=bins, range=(range_min, range_max))
bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

# ============================================================
# Select fitting region
# ============================================================
fit_mask = (bin_centers >= 0.05) & (bin_centers <= 10)
x_fit_all = bin_centers[fit_mask]
y_fit_all = counts[fit_mask]

# Skip zero bins (avoid div by 0 in weights)
nz = y_fit_all > 0
x_fit = x_fit_all[nz]
y_fit = y_fit_all[nz]
sigma = np.sqrt(y_fit)  # Poisson errors

# ============================================================
# Initial guesses
# ============================================================
N1_0 = float(np.max(y_fit)) if y_fit.size else 1.0
tau1_0 = 2.0
N2_0 = max(N1_0 / 10.0, 1.0)

# ============================================================
# Weighted least squares fit
# ============================================================
popt, pcov = curve_fit(
    exp_decay, x_fit, y_fit,
    p0=[N1_0, tau1_0, N2_0],
    sigma=sigma,
    absolute_sigma=True,
    bounds=([0.0, 0.0, 0.0], [np.inf, np.inf, np.inf]),
    maxfev=20000,
)
N1_fit, tau1_fit, N2_fit = popt
perr = np.sqrt(np.diag(pcov))
N1_err, tau1_err, N2_err = perr

# ============================================================
# Chi² / ndf
# ============================================================
y_exp = exp_decay(x_fit, *popt)
residual = (y_fit - y_exp) / sigma
chi2 = np.sum(residual**2)
ndf = max(len(x_fit) - len(popt), 1)
chi2_ndf = chi2 / ndf

# ============================================================
# Smooth curve for plotting
# ============================================================
x_smooth = np.linspace(0.05, 10, 500)
y_smooth = exp_decay(x_smooth, *popt)

# ============================================================
# Plot histogram
# ============================================================
plt.hist(TPC_lifetime, bins=bins, range=(range_min, range_max),
         edgecolor='magenta', histtype='step', linewidth=0.8, label="MC Data")

# Plot fitted curve
plt.plot(
    x_smooth, y_smooth, 'b-',
    label=(
        r'Fit: $N_1 e^{-t/\tau_1} + N_2 e^{-t/2.1}$' + '\n'
        + rf'$N_1={N1_fit:.1f}\pm{N1_err:.1f}$,  $\tau_1={tau1_fit:.3f}\pm{tau1_err:.3f}\ \mu s$' + '\n'
        + rf'$N_2={N2_fit:.1f}\pm{N2_err:.1f}$,  $\tau_2=2.1\ \mu s$ (fixed)' + '\n'
        + rf'$\chi^2/\mathrm{{ndf}} = {chi2:.1f}/{ndf}$'
    )
)

# ============================================================
# Style
# ============================================================
plt.xlim(-1, 18)
plt.ylim(0.1, 200000)
plt.yscale('log')
plt.xlabel('Lifetime [$\mu$s]', labelpad=2, fontsize=8)
plt.ylabel('#Counts', labelpad=1, fontsize=8)
plt.tick_params(labelsize=7)
plt.grid(color='gray', linestyle='--', linewidth=0.2, zorder=0)
plt.title('Lifetime Distribution of Michel True Events', fontsize=9)
plt.legend(fontsize=6)
plt.tight_layout()
plt.show()

# ============================================================
# Print results
# ============================================================
print(
    f"N1 = {N1_fit:.3f} ± {N1_err:.3f}\n"
    f"tau1 = {tau1_fit:.5f} ± {tau1_err:.5f} us\n"
    f"N2 = {N2_fit:.3f} ± {N2_err:.3f}\n"
    f"chi2/ndf = {chi2:.2f}/{ndf} = {chi2_ndf:.3f}"
)













#Fitting seperating for Michel e^+ and e^-=============================================================

# --- single exponential ---
def exp1(t, N, tau):
    return N * np.exp(-t / tau)

# --- plotting / binning setup ---
plt.figure(figsize=(4, 3), dpi=400)
bin_width = 0.05
range_min, range_max = -20, 20
bins = np.arange(range_min, range_max + bin_width, bin_width)

# --- helper: weighted fit (σ = sqrt(y)), skip zeros, draw ---
def fit_and_draw(sample, label_prefix, fit_range=(0.05, 10)):
    # histogram
    counts, bin_edges = np.histogram(sample, bins=bins, range=(range_min, range_max))
    bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

    # choose fit window & drop zero bins
    m = (bin_centers >= fit_range[0]) & (bin_centers <= fit_range[1])
    x_all = bin_centers[m]
    y_all = counts[m]
    nz = y_all > 0
    x_fit = x_all[nz]
    y_fit = y_all[nz]

    # abort gracefully if too few points
    if len(y_fit) < 3:
        plt.hist(sample, bins=bins, range=(range_min, range_max),
                 histtype='step', linewidth=0.8, label=f"{label_prefix} MC data")
        return (np.nan, np.nan, np.nan, np.nan, np.nan)

    sigma = np.sqrt(y_fit)

    # safe initial guesses & bounds
    N0, tau0 = float(np.max(y_fit)), 2.0
    bounds = ([0.0, 1e-6], [np.inf, np.inf])

    # weighted least squares
    popt, pcov = curve_fit(
        exp1, x_fit, y_fit, p0=[N0, tau0],
        sigma=sigma, absolute_sigma=True,
        bounds=bounds, maxfev=20000
    )
    N_fit, tau_fit = popt
    N_err, tau_err = np.sqrt(np.diag(pcov))

    # chi2/ndf
    y_mod = exp1(x_fit, *popt)
    r = (y_fit - y_mod) / sigma
    chi2 = float(np.sum(r**2))
    ndf = max(len(x_fit) - len(popt), 1)
    chi2_ndf = chi2 / ndf

    # smooth curve
    x_smooth = np.linspace(fit_range[0], fit_range[1], 500)
    y_smooth = exp1(x_smooth, *popt)

    # draw histogram (step)
    plt.hist(sample, bins=bins, range=(range_min, range_max),
             histtype='step', linewidth=0.8, label=f"{label_prefix} data")

    # draw fit
    plt.plot(
        x_smooth, y_smooth,
        label=(fr"{label_prefix} fit: $Ne^{{-t/\tau}}$" + "\n"
               + rf"$N={N_fit:.1f}\pm{N_err:.1f}$,  $\tau={tau_fit:.3f}\pm{tau_err:.3f}\ \mu s$" + "\n"
               + rf"$\chi^2/\mathrm{{ndf}} = {chi2:.1f}/{ndf}$")
    )

    return (N_fit, tau_fit, N_err, tau_err, chi2_ndf)

# ---- run for both lists ----
N_p, tau_p, dN_p, dtau_p, chi2ndf_p = fit_and_draw(np.asarray(TPC_life_pM), "e⁺")
N_n, tau_n, dN_n, dtau_n, chi2ndf_n = fit_and_draw(np.asarray(TPC_life_nM), "e⁻")

# --- final plot cosmetics (kept similar to your style) ---
plt.xlim(-2, 18)
plt.ylim(0.1, 1000)
plt.yscale('log')
plt.xlabel('Lifetime [$\mu$s]', labelpad=2, fontsize=8)
plt.ylabel('#Counts', labelpad=1, fontsize=8)
plt.tick_params(labelsize=7)
plt.grid(color='gray', linestyle='--', linewidth=0.2, zorder=0)
plt.title('Lifetime Distributions: Michel $e^+$ vs $e^-$', fontsize=9)
plt.legend(fontsize=6)
plt.tight_layout()
plt.show()
```
