```python
import numpy as np
import matplotlib.pyplot as plt

# Data
MS_cut = [0.4, 0.5, 0.6, 0.7, 0.8]

statNum = [5902, 4457, 3706, 2908, 1875]
purity =  [90.3, 92.7, 94.4, 95.7, 96.5]

tau_fit = [1885, 1902, 1907, 1905, 1875]
tau_unc = [41,   45,   50,   55,   68]

chi2 = [36.8, 34.5, 31.9, 39.6, 47.4]
ndf = 27
chi2_ndf = [x/ndf for x in chi2]   # compute chi2/ndf


# ============================================================
# Plot 1: statNum (left) + purity (right)
# ============================================================

fig, ax1 = plt.subplots(figsize=(5,4), dpi=400)

# --- Left axis: statNum ---
ax1.plot(MS_cut, statNum, 'o', color='black', markersize=4, label='Statistics')
ax1.set_xlabel("Michel Score Cut")
ax1.set_ylabel("Statistical Count", color='black')
ax1.tick_params(axis='y', labelcolor='black')
ax1.set_xlim(0.3, 0.9)
ax1.set_ylim(0, 10000)
ax1.grid(color='gray', linestyle='--', linewidth=0.2)

# --- Right axis: Purity ---
ax2 = ax1.twinx()
ax2.plot(MS_cut, purity, 's--', color='blue', markersize=4, label='Purity (%)')
ax2.set_ylabel("Purity (%)", color='blue')
ax2.tick_params(axis='y', labelcolor='blue')
ax2.set_ylim(0, 120)

# --- Combined legend ---
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper right", fontsize=9)

plt.show()



# ============================================================
# Plot 2: tau_fit with uncertainty (left) + chi2/ndf (right)
# ============================================================

fig, ax1 = plt.subplots(figsize=(5,4), dpi=400)

# --- Left: tau fit with errors ---
ax1.errorbar(
    MS_cut, tau_fit, yerr=tau_unc,
    fmt='o', color='red', capsize=3, markersize=4, label=r'$\tau_{\rm fit}$'
)
ax1.axhline(1910, color='magenta', linestyle='--', linewidth=1,
            label="Expected Lifetime")

ax1.set_xlabel("Michel Score Cut")
ax1.set_ylabel(r"Fitted Lifetime $\tau_{\rm fit}$ (ns)", color='red')
ax1.tick_params(axis='y', labelcolor='red')
ax1.set_xlim(0.3, 0.9)
ax1.set_ylim(1800, 2300)
ax1.grid(color='gray', linestyle='--', linewidth=0.2)

# --- Right: chi2/ndf ---
ax2 = ax1.twinx()
ax2.plot(
    MS_cut, chi2_ndf, 'd--', color='green', markersize=4,
    label=r'$\chi^2/\mathrm{ndf}$'
)

# --- Add χ²/ndf band on right y-axis ---
ax2.axhspan(
    0.73, 1.27,
    facecolor='lightgreen',
    alpha=0.3,
    zorder=0,
    label=r'1$\sigma$ region of $\chi^2/\mathrm{ndf}$'
)

ax2.set_ylabel(r'$\chi^2/\mathrm{ndf}$', color='green')
ax2.tick_params(axis='y', labelcolor='green')
ax2.set_ylim(0, 2)

# Combined legend
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper left", fontsize=9)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
MS_cut = [0.4, 0.5, 0.6, 0.7, 0.8]

statNum = [5902, 4457, 3706, 2908, 1875]
purity = [90.3, 92.7, 94.4, 95.7, 96.5]

tau_fit = [1885, 1902, 1907, 2908, 1875]
tau_unc = [41, 45, 50, 55, 68]

chi2 = [36.8, 34.5 31.9, 39.6, 47.4]
ndf = 27



#Plot 1: Stat & purity ========================================
# Figure setup
fig, ax1 = plt.subplots(figsize=(5, 4), dpi=400)

# --- Left y-axis: tau fit  with uncertainties ---
ax1.errorbar(
    MS_cut, tau_fit, yerr=tau_unc,
    fmt='o', color='blue', ecolor='blue',
    elinewidth=1, capsize=2, markersize=4, label=r'$\tau_{\rm fit}$'
)

ax1.axhline(
    y=1910, color='magenta', linestyle='--', linewidth=1,
    label=r'Theoretical Expected lifetime'
)


ax1.set_xlabel('Michel Score Cut')
ax1.set_ylabel(r'Fitted Lifetime $\tau_{\rm fit}$ (ns)', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')
ax1.set_xlim(0.3, 0.9)
ax1.set_ylim(1500, 2200)

plt.grid(color='gray', linestyle='--', linewidth=0.2, zorder=0)

# --- Right y-axis: purity ---
ax2 = ax1.twinx()
ax2.plot(
    MS_cut, purity, 's--', color='tab:red', markersize=4, label='Purity'
)
ax2.set_ylabel('Purity (%)', color='tab:red')
ax2.tick_params(axis='y', labelcolor='tab:red')
ax2.set_ylim(0, 120)

# --- Combined legend ---
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left', fontsize=7)



#plt.tight_layout()
plt.show()
```
