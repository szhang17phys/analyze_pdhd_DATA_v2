```python
#Decay points distribution:==========================================
print("Decay points distribution: ")

decayX_update_2 = np.array(decayX_update_2)
neg_X = np.sum(decayX_update_2 < 0)
pos_X = np.sum(decayX_update_2 >= 0)
print("\n# of X < 0 : ", neg_X, "({:.2f}%)".format(neg_X/len(decayX_update_2)*100))
print("# of X >= 0: ", pos_X, "({:.2f}%)".format(pos_X/len(decayX_update_2)*100))


# Plotting the histogram (X distribution)---------------
plt.figure(figsize=(4,3),dpi=400)
#APA planes---
plt.plot([-356, -356], [0, 2000], color="blue", linewidth=0.8, linestyle='--')
plt.plot([356, 356], [0, 2000], color="blue", linewidth=0.8, linestyle='--')
#Define the range and bin width
bin_width = 20
range_min = -400
range_max = 400
bins = np.arange(range_min, range_max + bin_width, bin_width)

plt.hist(decayX_update_2, bins=bins, range=(range_min, range_max), edgecolor='magenta', 
         histtype='step', linewidth=0.8)

plt.xlim(-450, 450)
plt.xlabel(r'$x_1$ [cm]', labelpad=2, fontsize=8)
plt.ylabel('#Counts', labelpad=1, fontsize=8)
plt.gca().invert_xaxis()  # Flip x-axis
plt.ylim(0.1, 500)
#plt.yscale('log')
plt.tick_params(labelsize=7)#set the size of ticks---
plt.grid(color='gray', linestyle='--', linewidth=0.2, zorder=0)
plt.show()  


# Compute histogram and print bin entries as a list
hist_counts, bin_edges = np.histogram(decayX_update_2, bins=bins, range=(range_min, range_max))
print(hist_counts.tolist())




#Y-Z plane ploting (Side View)----------------------------------------
plt.figure(figsize=(4,3),dpi=400)
#auxiliary line, along y(geo) and x axis---
aux2 = [cageMinY, cageMaxY, cageMaxY, cageMinY, cageMinY] #y shoud be y---
aux1 = [cageMinZ, cageMinZ, cageMaxZ, cageMaxZ, cageMinZ]
plt.plot(aux1, aux2, color="blue", linewidth=0.8, linestyle='-')

#Drawing of data points------
plt.scatter(decayZ_update_2, decayY_update_2, c='magenta', s=2.0, alpha=1.0, marker='.', label='Magenta Bias')

plt.axis('scaled')
plt.xlim(-50, 515)#z---
plt.ylim(-50, 660)
plt.xlabel('Z axis [cm]', labelpad=1, fontsize=5)
plt.ylabel('Y axis [cm]', labelpad=1, fontsize=5)
plt.tick_params(labelsize=4)#set the size of ticks---
plt.grid(color='gray', linestyle='--', linewidth=0.3, zorder=0)
plt.show()  





#-------------------------------------------------------------------
#Divide decay points in four APA chambers---------------------------
decayX_apa1 = []
decayX_apa2 = []
decayX_apa3 = []
decayX_apa4 = []

decayZ_apa1 = []
decayZ_apa2 = []
decayZ_apa3 = []
decayZ_apa4 = []

for i in range(0, len(decayZ_update_2)):
    z = decayZ_update_2[i]
    x = decayX_update_2[i]

    if x <= 0 and z <= 231.266: #APA1
        decayX_apa1.append(x)
        decayZ_apa1.append(z)              
    elif x <= 0 and z > 231.266: #APA2
        decayX_apa2.append(x)
        decayZ_apa2.append(z)       
    elif x > 0 and z <= 231.266: #APA3
        decayX_apa3.append(x)
        decayZ_apa3.append(z) 
    elif x > 0 and z > 231.266: #APA4
        decayX_apa4.append(x)
        decayZ_apa4.append(z) 



print("\n#decayX_TPC (total): ", len(decayX_update_2))
print("#decayX_apa1: ", len(decayX_apa1), "({:.2f}%)".format(len(decayX_apa1)/len(decayX_update_2)*100))
print("#decayX_apa2: ", len(decayX_apa2), "({:.2f}%)".format(len(decayX_apa2)/len(decayX_update_2)*100))
print("#decayX_apa3: ", len(decayX_apa3), "({:.2f}%)".format(len(decayX_apa3)/len(decayX_update_2)*100))
print("#decayX_apa4: ", len(decayX_apa4), "({:.2f}%)".format(len(decayX_apa4)/len(decayX_update_2)*100))






#Z-X plane ploting (Top View)----------------------------------------
plt.figure(figsize=(4,3),dpi=400)
#auxiliary line, along y(geo) and x axis---
aux1 = [0, 465, 465, 0, 0] #y shoud be y---
aux2 = [356, 356, -356, -356, 356]
plt.plot(aux1, aux2, color="blue", linewidth=0.8, linestyle='-')

# Create a custom colormap where zero-frequency regions are white
plasma = plt.cm.plasma.copy()
plasma.set_under('white')  # Sets zero values to white

# Create a 2D histogram
hist, xedges, yedges, img = plt.hist2d(decayZ_update_2, decayX_update_2, bins=(120, 120), range=[[zmin, zmax], [xmin, xmax]],
                                       cmap=plasma, cmin=1)  # cmin=1 ensures zero bins remain white

# Add color bar
cbar = plt.colorbar(pad=-0.3)
cbar.set_label('Counts', fontsize=6)
cbar.ax.tick_params(labelsize=5)

plt.axis('scaled')
plt.ylim(-450, 450)#z---
plt.xlim(-50, 515)
plt.ylabel('X axis [cm]', labelpad=1, fontsize=5)
plt.xlabel('Z axis [cm]', labelpad=1, fontsize=5)
plt.title('Top View', fontsize=7)
plt.tick_params(labelsize=4)#set the size of ticks---
plt.grid(color='gray', linestyle='--', linewidth=0.3, zorder=0)
plt.show()






#Y-X plane ploting (Front View)----------------------------------------
plt.figure(figsize=(4,3),dpi=400)
#auxiliary line, along y(geo) and x axis---
aux2 = [0, 610, 610, 0, 0] #y shoud be y---
aux1 = [356, 356, -356, -356, 356]
plt.plot(aux1, aux2, color="blue", linewidth=0.8, linestyle='-')

#Drawing of data points------
plt.scatter(decayX_update_2, decayY_update_2, c='magenta', s=2.0, alpha=1.0, marker='.', label='Magenta Bias')

plt.axis('scaled')
plt.xlim(-450, 450)#z---
plt.ylim(-50, 660)
plt.xlabel('X axis [cm]', labelpad=1, fontsize=5)
plt.ylabel('Y axis [cm]', labelpad=1, fontsize=5)
plt.tick_params(labelsize=4)#set the size of ticks---
plt.grid(color='gray', linestyle='--', linewidth=0.3, zorder=0)
plt.show() 












# Muon Lifetime distribution (NO fitting)--------------------------------------------
plt.figure(figsize=(4, 3), dpi=400)

# Define histogram parameters
bin_width = 500
range_min = 500
range_max = 16000
bins = np.arange(range_min, range_max + bin_width, bin_width)

# Plot histogram-----------------
plt.hist(mu_lifetime_ns_update_2, bins=bins, range=(range_min, range_max),
         edgecolor='blue', histtype='step', linewidth=0.8,
         label='Muon Lifetime')

plt.xlim(0, 16000)
plt.ylim(0, 1200)
plt.xlabel('Time [ns]', fontsize=8)
plt.ylabel('#Counts', fontsize=8)
plt.tick_params(labelsize=7)
plt.grid(color='gray', linestyle='--', linewidth=0.2, zorder=0)
plt.legend(loc='upper center', bbox_to_anchor=(0.78, 0.95), fontsize=11, markerscale=5, ncol=1, prop={'size':8})
plt.show()







#Fitting---------------------------------------------------------------------------
plt.figure(figsize=(4, 3), dpi=400)

bin_width = 500
range_min = 500
range_max = 15500
bins = np.arange(range_min, range_max + bin_width, bin_width)

n, bin_edges, _ = plt.hist(
    mu_lifetime_ns_update_2,
    bins=bins, range=(range_min, range_max),
    edgecolor='blue', histtype='step', linewidth=0.8,
    label='Muon Lifetime Preliminary'
)

bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])
yerr = np.sqrt(n)

# ------------------ masked fitting (n > 0) ------------------
mask = n > 0
x_fit = bin_centers[mask]
y_fit = n[mask]
sigma_fit = yerr[mask]

def exp_func(t, N0, tau0, C):
    return N0 * np.exp(-t / tau0) + C

# initial guesses & bounds (positive params)
p0 = [y_fit.max(), 2000.0, 1.0]
bounds = ([0.0, 1.0, 0.0], [np.inf, np.inf, np.inf])

popt, pcov = curve_fit(
    exp_func, x_fit, y_fit, p0=p0,
    sigma=sigma_fit, absolute_sigma=True, bounds=bounds
)
perr = np.sqrt(np.diag(pcov))

# ------------------ chi² using safe errors ------------------
yerr_plot = np.where(n == 0, 1.0, yerr)
expected_all = exp_func(bin_centers, *popt)
residuals = n - expected_all
chi2 = np.sum((residuals / yerr_plot) ** 2)
ndf = np.count_nonzero(mask) - len(popt)
chi2_ndf = chi2 / ndf

# ------------------ printing ------------------
print("\nBin information:")
for i in range(len(n)):
    print(f"Bin {i+1}: Range = [{bin_edges[i]:.0f}, {bin_edges[i+1]:.0f}) | "
          f"Center = {bin_centers[i]:.0f} ns | "
          f"Entries = {n[i]:.0f} ± {yerr[i]:.1f} | "
          f"Expected = {expected_all[i]:.2f} | "
          f"Residual = {residuals[i]:.2f} | "
          f"Residual / σ = {residuals[i]/yerr_plot[i]:.2f}")

print("\nFitted parameters (masked least squares):")
print(f"N0   = {popt[0]:.2f} ± {perr[0]:.2f}")
print(f"tau0 = {popt[1]:.2f} ± {perr[1]:.2f} ns")
print(f"C    = {popt[2]:.2f} ± {perr[2]:.2f}")
print(f"[Goodness of Fit] χ² / ndf = {chi2:.1f} / {ndf} = {chi2_ndf:.2f}")

# ------------------ plotting ------------------
plt.errorbar(bin_centers, n, yerr=yerr_plot, fmt='o', color='blue', markersize=2)

t_fit = np.linspace(x_fit.min(), x_fit.max(), 1000)
fit_line, = plt.plot(t_fit, exp_func(t_fit, *popt), color='red', linestyle='--',
                     label=f'τ = {popt[1]:.0f} ± {perr[1]:.0f} ns')

# text-only legend entries
n0_text  = f'N₀ = {popt[0]:.0f} ± {perr[0]:.0f}'
c_text   = f'C = {popt[2]:.1f} ± {perr[2]:.1f}'
chi_text = f'χ² / ndf = {chi2:.1f} / {ndf}'

dummy_n0  = plt.Line2D([], [], color='none', label=n0_text)
dummy_c   = plt.Line2D([], [], color='none', label=c_text)
dummy_chi = plt.Line2D([], [], color='none', label=chi_text)
hist_handle = plt.Line2D([], [], color='blue', linestyle='-', label='Muon Lifetime Preliminary')

plt.legend(handles=[hist_handle, fit_line, dummy_n0, dummy_c, dummy_chi],
           loc='upper center', bbox_to_anchor=(0.7, 0.99),
           fontsize=7, ncol=1, prop={'size': 6})

plt.xlim(0, 16000)
plt.xlabel('Time [ns]', labelpad=2, fontsize=8)
plt.ylabel('#Counts', labelpad=1, fontsize=8)
plt.tick_params(labelsize=7)
plt.grid(color='gray', linestyle='--', linewidth=0.2, zorder=0)
plt.title('Muon Lifetime Fit (masked least squares)', fontsize=8)
plt.show()


```
