###  SPE template drawing:

#### SPE_NP04_FBK_2024_without_pretrigger.dat, SPE_NP04_HPK_2024_without_pretrigger.dat

```python
import matplotlib.pyplot as plt

# Specify the file name
file_FBK = './templates_2024/SPE_NP04_FBK_2024_without_pretrigger.dat'
file_HPK = './templates_2024/SPE_NP04_HPK_2024_without_pretrigger.dat'

# Read the data from the file
values = []
with open(file_FBK, 'r') as file:
    for line in file:
        try:
            values.append(float(line.strip()))  # Convert each line to a float
        except ValueError:
            print(f"Skipping invalid line: {line.strip()}")

# Ensure there are 512 values in the file
if len(values) != 512:
    print(f"Warning: Expected 512 values but found {len(values)}.")

# Read the data from the file
valuesH = []
with open(file_HPK, 'r') as file:
    for line in file:
        try:
            valuesH.append(float(line.strip()))  # Convert each line to a float
        except ValueError:
            print(f"Skipping invalid line: {line.strip()}")

# Ensure there are 512 values in the file
if len(valuesH) != 512:
    print(f"Warning: Expected 512 values but found {len(valuesH)}.")


# Create the x values (1 to 512)
x_values = list(range(1, len(values) + 1))

# Plot the data----------
plt.figure(figsize=(4,3),dpi=300)
plt.plot(x_values, values, marker='s', linestyle='-', color='b', markersize=0.8, linewidth=1, label='SPE_NP04_FBK_2024_without_pretrigger.dat')
plt.plot(x_values, valuesH, marker='s', linestyle='-', color='r', markersize=0.8, linewidth=1, label='SPE_NP04_HPK_2024_without_pretrigger.dat')

plt.xlabel('Ticks', labelpad=1, fontsize=8)
plt.ylabel('ADC', labelpad=1, fontsize=8)
plt.tick_params(labelsize=7)#set the size of ticks---

plt.title("SPE Templates", fontsize=10)
plt.legend(loc='upper center', bbox_to_anchor=(0.6, 0.95), fontsize=8, markerscale=3, ncol=1, prop={'size': 6})
plt.grid(color='gray', linestyle='--', linewidth=0.5, zorder=0)
plt.show()

```

```python
import matplotlib.pyplot as plt
import os

# Set directory containing .txt files
directory = './templates_2025/v1'

# List all .txt files
all_files = [os.path.join(directory, f) for f in os.listdir(directory) if f.endswith('.txt')]

# Sort files for consistent order
all_files.sort()

# Plot setup
plt.figure(figsize=(4, 3), dpi=300)

# Loop over all files
for filepath in all_files:
    values = []
    with open(filepath, 'r') as file:
        for line in file:
            try:
                values.append(float(line.strip()))
            except ValueError:
                continue

    if len(values) != 512 and len(values) != 1024:
        print(f"Warning: {os.path.basename(filepath)} has {len(values)} values (expected 512 or 1024)")

    # Compute max position
    if values:
        max_index = values.index(max(values)) + 1  # +1 to match tick index starting from 1
        print(f"{os.path.basename(filepath)} --> Max at tick {max_index} (value = {max(values):.2f})")

    # Plot
    x_values = list(range(1, len(values) + 1))
    label = os.path.basename(filepath)
    plt.plot(x_values, values, linewidth=0.8, markersize=0.5, label=label)

# Plot decoration
# plt.xlim(50, 200)
plt.xlabel('Ticks', labelpad=1, fontsize=8)
plt.ylabel('ADC', labelpad=1, fontsize=8)
plt.tick_params(labelsize=7)
plt.title("All SPE Templates", fontsize=10)

# plt.legend(fontsize=5, loc='center left', bbox_to_anchor=(1, 0.5), ncol=1)
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import os

# Set directory containing .txt files
directory = './templates_2025/v0'

# List all .txt files
all_files = [os.path.join(directory, f) for f in os.listdir(directory) if f.endswith('.txt')]
all_files.sort()

# Split into fbk and hpk groups
fbk_files = [f for f in all_files if 'fbk' in f.lower()]
hpk_files = [f for f in all_files if 'hpk' in f.lower()]

def plot_group(file_list, group_name):
    plt.figure(figsize=(4, 3), dpi=300)
    for filepath in file_list:
        values = []
        with open(filepath, 'r') as file:
            for line in file:
                try:
                    values.append(float(line.strip()))
                except ValueError:
                    continue

        if len(values) != 512 and len(values) != 1024:
            print(f"Warning: {os.path.basename(filepath)} has {len(values)} values (expected 512 or 1024)")

        if values:
            max_index = values.index(max(values)) + 1
            print(f"{os.path.basename(filepath)} --> Max at tick {max_index} (value = {max(values):.2f})")

        x_values = list(range(1, len(values) + 1))
        label = os.path.basename(filepath)
        plt.plot(x_values, values, linewidth=0.8, markersize=0.5, label=label)

    plt.xlabel('Ticks', labelpad=1, fontsize=8)
    plt.ylabel('ADC', labelpad=1, fontsize=8)
    plt.tick_params(labelsize=7)
    plt.title(f"{group_name.upper()} SPE Templates", fontsize=10)
    # Optional legend
    # plt.legend(fontsize=5, loc='center left', bbox_to_anchor=(1, 0.5), ncol=1)
    plt.grid(color='gray', linestyle='--', linewidth=0.5)
    plt.tight_layout()
    plt.show()

# Plot both groups
plot_group(fbk_files, "FBK")
plot_group(hpk_files, "HPK")
```

## RSL vs Energy and Wavelength:

Based on the data shared by Jose.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import interp1d


# Sample data---
energy = [1.18626, 1.68626, 2.18626, 2.68626, 3.18626, 3.68626, 4.18626, 4.68626, 5.18626, 5.68626, 
            6.18626, 6.68626, 7.18626, 7.68626, 8.18626, 8.68626, 9.18626, 9.68626, 10.1863, 10.6863, 11.1863]
rsl = [1200800, 390747, 128633, 54969.1, 27191.8, 14853.7, 8716.9, 5397.42, 3481.37, 2316.51,
       1577.63, 1092.02, 763.045, 534.232, 371.335, 252.942, 165.38, 99.9003, 51.2653, 17.495, 0.964341]
#Wavelength---
energy_array = np.array(energy)
wave_array_tmp = np.reciprocal(energy_array)#array---
wave_array = 2*np.pi*197.326980*wave_array_tmp
wave = wave_array.tolist()#list---




#RSL vs Energy Drawing------------------------------
plt.figure(figsize=(4,3),dpi=300)

plt.scatter(energy, rsl, label='Geant4 results', s=1.5, color='red', marker='s')

# Perform quadratic interpolation
quadratic_interp = interp1d(energy, rsl, kind='quadratic')

#Create a smooth line connecting the points using quadratic interpolation---
smooth_line = np.linspace(min(energy), max(energy), 1000)
smooth_line_y = quadratic_interp(smooth_line)
plt.plot(smooth_line, smooth_line_y, label='Quadratic Interpolation', color='blue', linewidth=0.5)


# Add labels and title---
plt.xlabel('E($\gamma$) [eV]', labelpad=1, fontsize=8)
plt.ylabel('RSL [cm]', labelpad=1, fontsize=8)
plt.tick_params(labelsize=7)#set the size of ticks---
plt.yscale('log')

plt.title("RSL Dependence on Photon Energy", fontsize=10)
plt.legend(loc='upper center', bbox_to_anchor=(0.5, 0.99), fontsize=6, markerscale=3, ncol=2, prop={'size': 4})
plt.grid(color='gray', linestyle='--', linewidth=0.5, zorder=0)
plt.show()






#RSL vs Energy Drawing------------------------------
plt.figure(figsize=(4,3),dpi=300)

plt.scatter(wave, rsl, label='Geant4 results', s=1.5, color='red', marker='s')

# Perform quadratic interpolation
quadratic_interp = interp1d(wave, rsl, kind='quadratic')

#Create a smooth line connecting the points using quadratic interpolation---
smooth_line = np.linspace(min(wave), max(wave), 1000)
smooth_line_y = quadratic_interp(smooth_line)
plt.plot(smooth_line, smooth_line_y, label='Quadratic Interpolation', color='blue', linewidth=0.5)


# Add labels and title---
plt.xlabel('Wavelength($\gamma$) [nm]', labelpad=1, fontsize=8)
plt.ylabel('RSL [cm]', labelpad=1, fontsize=8)
plt.tick_params(labelsize=7)#set the size of ticks---
plt.yscale('log')

plt.title("RSL Dependence on Photon Wavelength", fontsize=10)
plt.legend(loc='upper center', bbox_to_anchor=(0.5, 0.99), fontsize=6, markerscale=3, ncol=2, prop={'size': 4})
plt.grid(color='gray', linestyle='--', linewidth=0.5, zorder=0)
plt.show()


```
